```python
# ==========================================
# Modular Calculator Mini-Program
# ==========================================

# 1. Define functions for each operation using parameters and return values

def add_numbers(num1, num2):
    """Adds two numbers and returns the sum."""
    return num1 + num2

def subtract_numbers(num1, num2):
    """Subtracts the second number from the first and returns the difference."""
    return num1 - num2

def multiply_numbers(num1, num2):
    """Multiplies two numbers and returns the product."""
    return num1 * num2

def divide_numbers(num1, num2):
    """Divides the first number by the second and returns the quotient. Handles division by zero."""
    if num2 == 0:
        return "Cannot divide by zero."
    return num1 / num2


# 2. Main Program Flow

print("--- Welcome to the Modular Calculator ---")

# Ask the user to enter two numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Ask the user to choose an operation
print("\nChoose operation:")
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")

choice = input("\nEnter choice: ")

# Call the correct function, store the returned value, and display the result
if choice == '1':
    result = add_numbers(num1, num2)
    print(f"\nResult: {result}")
elif choice == '2':
    result = subtract_numbers(num1, num2)
    print(f"\nResult: {result}")
elif choice == '3':
    result = multiply_numbers(num1, num2)
    print(f"\nResult: {result}")
elif choice == '4':
    result = divide_numbers(num1, num2)
    print(f"\nResult: {result}")
else:
    print("\nInvalid choice! Please select a valid option from 1 to 4.")
```
